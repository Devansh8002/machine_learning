```python

# Python Assignment: Variance and Standard Deviation

import math
import numpy as np
import pandas as pd
import seaborn as sns


# Functions for calculations

def mean(data):
    return sum(data) / len(data)


def population_variance(data):
    avg = mean(data)
    total = 0

    for x in data:
        total += (x - avg) ** 2

    return total / len(data)


def sample_variance(data):
    avg = mean(data)
    total = 0

    for x in data:
        total += (x - avg) ** 2

    return total / (len(data) - 1)


def population_std(data):
    return math.sqrt(population_variance(data))


def sample_std(data):
    return math.sqrt(sample_variance(data))


def show_result(data):
    print("Mean:", round(mean(data), 2))
    print("Population Variance:", round(population_variance(data), 2))
    print("Sample Variance:", round(sample_variance(data), 2))
    print("Population Standard Deviation:",
          round(population_std(data), 2))
    print("Sample Standard Deviation:",
          round(sample_std(data), 2))


# Q1. Employee Performance

print("\n========== Q1: EMPLOYEE PERFORMANCE ==========")

scores = [45, 50, 52, 55, 58, 60, 65, 75]

show_result(scores)

print("\nExplanation:")
print("Population variance divides by n.")
print("Sample variance divides by n - 1.")
print("Sample variance is higher because it uses n - 1.")

print("Difference in variance:",
      round(sample_variance(scores) -
            population_variance(scores), 2))

print("For a sample of a larger workforce,")
print("sample variance should be used.")


# Q2. Customer Spending Analysis

print("\n========== Q2: CUSTOMER SPENDING ==========")

df = sns.load_dataset("tips")

for col in ["total_bill", "tip"]:
    data = df[col].dropna().tolist()

    avg = mean(data)
    var = population_variance(data)
    std = population_std(data)
    cv = (std / avg) * 100

    print("\nColumn:", col)
    print("Mean:", round(avg, 2))
    print("Population Variance:", round(var, 2))
    print("Population Standard Deviation:", round(std, 2))
    print("Coefficient of Variation:", round(cv, 2), "%")


# Compare the five largest bills

bills = df["total_bill"].tolist()

largest = sorted(bills)[-5:]

# Remove the five largest observations
sorted_bills = sorted(bills)
new_bills = sorted_bills[:-5]

print("\nFive largest bills:", largest)

print("\nOriginal bill statistics:")
print("Mean:", round(mean(bills), 2))
print("Variance:", round(population_variance(bills), 2))
print("Standard Deviation:", round(population_std(bills), 2))

print("\nAfter removing five largest bills:")
print("Mean:", round(mean(new_bills), 2))
print("Variance:", round(population_variance(new_bills), 2))
print("Standard Deviation:", round(population_std(new_bills), 2))

print("Variance is sensitive to extreme values")
print("because differences from the mean are squared.")


# Q3. Variance and Standard Deviation Calculator

print("\n========== Q3: CALCULATOR ==========")

data_1 = [10, 20, 30, 40, 50]
data_2 = [20, 20, 20, 20, 20]
data_3 = [5, 10, 15, 20, 100]
data_4 = [12, 14, 16, 18, 20, 22]

datasets = {
    "Data 1": data_1,
    "Data 2": data_2,
    "Data 3": data_3,
    "Data 4": data_4
}

for name, data in datasets.items():
    print("\n", name, ":", data)
    show_result(data)

print("\nData 2 has zero variability.")
print("Data 3 is most affected by the extreme value 100.")
print("Its population variance is 1250.")
print("Its population standard deviation is 35.36.")


# Q4. Mathematical Transformations

print("\n========== Q4: TRANSFORMATIONS ==========")

measurements = [12, 15, 18, 20, 25, 30, 40]

add_10 = []
multiply_4 = []
multiply_minus_2 = []

for x in measurements:
    add_10.append(x + 10)
    multiply_4.append(x * 4)
    multiply_minus_2.append(x * -2)

datasets = {
    "Original": measurements,
    "Add 10": add_10,
    "Multiply by 4": multiply_4,
    "Multiply by -2": multiply_minus_2
}

for name, data in datasets.items():
    print("\n", name, ":", data)
    print("Mean:", round(mean(data), 2))
    print("Population Variance:",
          round(population_variance(data), 2))
    print("Population Standard Deviation:",
          round(population_std(data), 2))

print("\nAdding 10 changes the mean but not the spread.")
print("Multiplying by 4 multiplies variance by 16")
print("and standard deviation by 4.")
print("Multiplying by -2 makes the mean negative,")
print("but variance and standard deviation stay non-negative.")


# Q5. Outlier Detection and Spread Analysis

print("\n========== Q5: OUTLIER ANALYSIS ==========")

sales = [
    12, 14, 15, 15, 16, 17, 18,
    19, 20, 21, 22, 24, 25, 120
]

print("\nOriginal sales:")
print("Mean:", round(mean(sales), 2))
print("Median:", round(float(np.median(sales)), 2))
print("Population Variance:",
      round(population_variance(sales), 2))
print("Population Standard Deviation:",
      round(population_std(sales), 2))

# Calculate quartiles and IQR
q1 = np.percentile(sales, 25)
q3 = np.percentile(sales, 75)

iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

print("\nQ1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower)
print("Upper Bound:", upper)

outliers = []

for x in sales:
    if x < lower or x > upper:
        outliers.append(x)

print("Potential Outliers:", outliers)

# Remove the flagged observation for comparison
new_sales = []

for x in sales:
    if x not in outliers:
        new_sales.append(x)

print("\nAfter removing the outlier:")
print("Mean:", round(mean(new_sales), 2))
print("Median:", round(float(np.median(new_sales)), 2))
print("Population Variance:",
      round(population_variance(new_sales), 2))
print("Population Standard Deviation:",
      round(population_std(new_sales), 2))

old_var = population_variance(sales)
new_var = population_variance(new_sales)

old_std = population_std(sales)
new_std = population_std(new_sales)

print("\nVariance change (%):",
      round((new_var - old_var) / old_var * 100, 2), "%")

print("Standard Deviation change (%):",
      round((new_std - old_std) / old_std * 100, 2), "%")

print("\nConclusion:")
print("The value 120 greatly increases the mean and spread.")
print("The median better represents typical daily sales.")
print("The outlier should be investigated before removal.")

print("\n========== ASSIGNMENT COMPLETED ==========")
```
