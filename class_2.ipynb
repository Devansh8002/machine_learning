```python
array=([12,12,13,14,15,16,17,17,18,19,20,21,22])
```

```python
def variance_std(*args):
    total=0
    for i in args:
        total+=(-np.mean(args))**2
    calculated_variance= total/len(args)
    return calculated_variance,calculated_variance**(1/2)
```

```python
variance_std(*array)
```
