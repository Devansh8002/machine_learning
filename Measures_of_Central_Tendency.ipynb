```python
salaries = [18, 20, 22, 22, 24, 25, 25, 26,
            28, 30, 32, 35, 38, 42, 150]
```

```python

# Q1. Employee Salary Analysis

def mean(data):
    return sum(data) / len(data)

def median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 1:
        return data[n // 2]
    else:
        return (data[n // 2 - 1] + data[n // 2]) / 2

def modes(data):
    counts = {}

    for x in data:
        if x in counts:
            counts[x] += 1
        else:
            counts[x] = 1

    highest = max(counts.values())

    if highest == 1:
        return []

    result = []
    for x in counts:
        if counts[x] == highest:
            result.append(x)

    return result


salaries = [18, 20, 22, 22, 24, 25, 25, 26,
            28, 30, 32, 35, 38, 42, 150]

print("Q1: Employee Salary Analysis")
print("Mean:", round(mean(salaries), 2))
print("Median:", median(salaries))
print("Modes:", modes(salaries))

new_salaries = []
for x in salaries:
    if x != 150:
        new_salaries.append(x)

print("\nAfter removing outlier:")
print("Mean:", round(mean(new_salaries), 2))
print("Median:", median(new_salaries))
print("Modes:", modes(new_salaries))


# Q2. Customer Spending Analysis

import seaborn as sns
import pandas as pd

df = sns.load_dataset("tips")

print("\nQ2: Customer Spending Analysis")

for col in ["total_bill", "tip"]:
    data = df[col].tolist()
    print("\n", col)
    print("Mean:", round(mean(data), 2))
    print("Median:", median(data))
    print("Modes:", modes(data))

for meal in ["Lunch", "Dinner"]:
    data = df[df["time"] == meal]["total_bill"].tolist()
    print("\n", meal)
    print("Mean:", round(mean(data), 2))
    print("Median:", median(data))

bills = df["total_bill"].tolist()
sorted_bills = sorted(bills)

# Remove the three largest bills
new_bills = sorted_bills[:-3]

print("\nAfter removing three largest bills:")
print("Original mean:", round(mean(bills), 2))
print("New mean:", round(mean(new_bills), 2))
print("Original median:", median(bills))
print("New median:", median(new_bills))


# Q3. Central Tendency Calculator

print("\nQ3: Central Tendency Calculator")

data1 = [12, 15, 15, 18, 20, 22, 25]
data2 = [5, 5, 10, 10, 15, 20]
data3 = [7, 14, 21, 28, 35]
data4 = [100, 100, 100, 100]
data5 = [1, 2, 3, 4, 5, 100]

datasets = [data1, data2, data3, data4, data5]

for i in range(len(datasets)):
    data = datasets[i]
    print("\nDataset", i + 1)
    print("Mean:", round(mean(data), 2))
    print("Median:", median(data))
    print("Modes:", modes(data))


# Q4. Statistical Transformations

print("\nQ4: Statistical Transformations")

orders = [1, 1, 1, 2, 2, 2, 2, 3, 3, 4,
          4, 5, 5, 6, 8, 10, 15, 20, 25, 80]

print("Original data:")
print("Mean:", round(mean(orders), 2))
print("Median:", median(orders))
print("Modes:", modes(orders))

new_orders = orders[:-1]

print("\nAfter removing largest value:")
print("Mean:", round(mean(new_orders), 2))
print("Median:", median(new_orders))
print("Modes:", modes(new_orders))

add20 = []
multiply3 = []

for x in orders:
    add20.append(x + 20)
    multiply3.append(x * 3)

print("\nAfter adding 20:")
print("Mean:", round(mean(add20), 2))
print("Median:", median(add20))
print("Modes:", modes(add20))

print("\nAfter multiplying by 3:")
print("Mean:", round(mean(multiply3), 2))
print("Median:", median(multiply3))
print("Modes:", modes(multiply3))


# Q5. Missing Values and Outliers

print("\nQ5: Missing Values and Outliers")

sales = [12, 15, 14, 16, None, 15, 18, 14,
         17, 16, None, 19, 15, 120, 14]

s = pd.Series(sales)

print("Missing values:", s.isna().sum())
print("Missing percentage:",
      round(s.isna().mean() * 100, 2))

clean_sales = s.dropna().tolist()

print("\nSales statistics:")
print("Mean:", round(mean(clean_sales), 2))
print("Median:", median(clean_sales))
print("Modes:", modes(clean_sales))

# Find outliers using IQR
sorted_sales = sorted(clean_sales)

def find_quartiles(data):
    n = len(data)
    lower_half = data[:n // 2]
    upper_half = data[(n + 1) // 2:]
    return median(lower_half), median(upper_half)

q1, q3 = find_quartiles(sorted_sales)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower bound:", lower)
print("Upper bound:", upper)

outliers = []

for x in clean_sales:
    if x < lower or x > upper:
        outliers.append(x)

print("Potential outliers:", outliers)

new_sales = []
for x in clean_sales:
    if x != 120:
        new_sales.append(x)

print("\nAfter removing 120:")
print("Mean:", round(mean(new_sales), 2))
print("Median:", median(new_sales))
print("Modes:", modes(new_sales))

print("\nAll questions completed.")
```
